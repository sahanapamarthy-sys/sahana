```python
import numpy as np
import matplotlib.pyplot as plt 
```

```python
data = np.loadtxt("../../course_assignments/homework7/star_data.txt", skiprows = 1) 
```

```python
print (data.shape)
print("There are", data.shape[0], "rows")
```

```python
magnitude = data[:,0]
print(magnitude)
temperature = data[:,1]
print(temperature)

print("min magnitude:", np.min(magnitude))
print("max magnitude:", np.max(magnitude))

print("min temperature:", np.min(temperature))
print("max temperature:", np.max(temperature))

brightest = np.argmin(magnitude)
print(brightest)
dimmest = np.argmax(magnitude)
print(dimmest)

hottest = np.argmax(temperature)
print(hottest)
coldest = np.argmin(temperature)
print(coldest)
```

```python
log_magnitude = np.log(magnitude)
log_temperature = np.log(temperature)

plt.figure(figsize=(8, 6))
plt.scatter(log_temperature, log_magnitude, label="Stars")

plt.gca().invert_xaxis()
plt.gca().invert_yaxis()

plt.scatter(log_temperature[brightest],
            log_magnitude[brightest],
            color="black", marker="*", s=300, label="Brightest")
plt.scatter(log_temperature[dimmest],
            log_magnitude[dimmest],
            color="purple", marker="*", s=300, label="Dimmest")
plt.scatter(log_temperature[hottest],
            log_magnitude[hottest],
            color="green", marker="*", s=300, label="Hottest")
plt.scatter(log_temperature[coldest],
            log_magnitude[coldest],
            color="orange", marker="*", s=300, label="Coldest")

plt.legend()
plt.title("Hertzsprung-Rusell Diagram")
plt.xlabel("Temperature : log K")
plt.ylabel("Magnitue")

plt.show()
```
