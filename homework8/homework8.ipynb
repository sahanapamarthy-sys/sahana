```python
import pandas as pd
```

```python
data = pd.read_csv("../course_assignments/homework8/state_data.csv")
```

```python
print(data.shape)
print(f'There are {data.shape[0]} rows and {data.shape[1]} columns')
print(data.head())
print(data.tail())
#The differeny columns represent state abbrevations, latitude(N-S position), longitude(E-W position), and the names of the state. 
#The first state is Alaska
#The last state is Wyoming 
#There are 52 entries 
#It includes states that are not formally recognized like Puerto Rico and DC. 
```

```python
medianlat = data['latitude'].median()
print(f'The median of lat is {medianlat}')
medianlon = data['longitude'].median()
print(f'The median of long is {medianlon}')
```

```python
data['lat_dist'] = abs(data['latitude']-medianlat)
data['lon_dist'] = abs(data['latitude']-medianlon)
print(data.shape)
#There are 52 rows and 6 columns 
```

```python
closestlat = data.loc[data['lat_dist'].idxmin()]
farthestlat = data.loc[data['lat_dist'].idxmax()]
closestlon = data.loc[data['lon_dist'].idxmin()]
farthestlon = data.loc[data['lon_dist'].idxmax()]

print(f"Closest to median latitude: {closestlat['name']} ({closestlat['lat_dist']:.2f})")
print(f"Farthest from median latitude: {farthestlat['name']} ({farthestlat['lat_dist']:.2f})")

print(f"Closest to median longitude: {closestlon['name']} ({closestlon['lon_dist']:.2f})")
print(f"Farthest from median longitude: {farthestlon['name']} ({farthestlon['lon_dist']:.2f})")
```

```python
import seaborn as sns
import matplotlib.pyplot as plt 
```

```python
data2 = sns.load_dataset('tips')
data2 = data2.dropna()
```

```python
print(data2.shape)
print(data2.head)
print(data2.tail())
print(data2.columns)
```

```python
fig, axes = plt.subplots(2, 2, figsize=(10, 8))


axes[0, 0].scatter(data2["total_bill"], data2["tip"], color="blue")
axes[0, 0].set_title("Total Bill vs Tip")
axes[0, 0].set_xlabel("Total Bill ($)")
axes[0, 0].set_ylabel("Tip ($)")


axes[0, 1].scatter(data2["size"], data2["total_bill"], color="green")
axes[0, 1].set_title("Party Size vs Total Bill")
axes[0, 1].set_xlabel("Party Size")
axes[0, 1].set_ylabel("Total Bill ($)")


sns.histplot(data2["total_bill"], ax=axes[1, 0], color="purple")
axes[1, 0].set_title("Distribution of Total Bill")


sns.boxplot(x=data2["day"], y=data2["tip"], ax=axes[1, 1])
axes[1, 1].set_title("Tips by Day")

plt.tight_layout()
plt.show()
```

```python
#The first scatterplot shows how tips increase as the total bill increases.
#The second scatterplot shows how total bill changes with party size 
#The histogram shows the distribution of total bills amounts 
#the boxplot compares tips thoroughtout the different days 
#You understand the relationships and distributions through these plots. 
```
